```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
olympic_data = pd.read_csv('/content/athlete_events.csv')
noc = pd.read_csv('/content/noc_regions.csv')
merged_data = pd.merge(olympic_data, noc, on='NOC', how='left')
print(merged_data.isnull().sum())
print(merged_data.shape)
merged_data.head()
```

```python
gold_medalist = merged_data[merged_data['Medal'] == 'Gold']
print(gold_medalist.shape)
gold_medalist.head(5)
```

```python
print(merged_data['Medal'].unique())
```

```python
country_gold = gold_medalist.groupby('region').size()
Top_10_country = country_gold.sort_values(ascending=False).head(10)
print(Top_10_country)


Top_10_country.plot(
    kind="barh",
    figsize=(12, 8),
    title="Top 10 Countries with Gold Medals"
)
ax = plt.gca()
ax.bar_label(ax.containers[0])

plt.xlabel("Number of Gold Medals", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.xticks(fontsize=10)
plt.yticks(fontsize=10)
plt.tight_layout()
plt.gca().invert_yaxis()
plt.show()
```

```python
silver_medalist = merged_data[merged_data['Medal'] == "Silver"]
print(silver_medalist.shape)
silver_medalist.head(5)
```

```python
country_silver = silver_medalist.groupby('region').size()
Top_10_country_by_silver = country_silver.sort_values(ascending=False).head(10)
print(Top_10_country_by_silver)

Top_10_country_by_silver.plot(
    kind="barh",
    figsize=(12, 8),
    title="Top 10 countries with Silver Medals"
)
axs = plt.gca()
axs.bar_label(axs.containers[0])
plt.xlabel("Number of Medals", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.xticks(fontsize=10)
plt.gca().invert_yaxis()
plt.tight_layout()

```

```python
bronze_medalist = merged_data[merged_data['Medal'] == 'Bronze']
print(bronze_medalist.shape)

country_bronze = bronze_medalist.groupby('region').size()
Top_10_country_by_bronze = country_bronze.sort_values(ascending=False).head(10)
print(Top_10_country_by_bronze)

Top_10_country_by_bronze.plot(
    kind="barh",
    figsize=(12, 8),
    title="Top 10 country by bronze"
)

axb = plt.gca()
axb.bar_label(axb.containers[0])
plt.xlabel("Number of Medals", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.xticks(fontsize=10)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

```

```python
country_medal = merged_data[merged_data['Medal'].notna()]
print(country_medal.shape)
country_medal = country_medal.groupby('region').size()
Top_10_country_by_medal = country_medal.sort_values(ascending=False).head(10)
print(Top_10_country_by_medal)

Top_10_country_by_medal.plot(
    kind="barh",
    figsize=(12, 8),
    title="Top 10 country by Medal"
)
axm = plt.gca()
axm.bar_label(axm.containers[0])
plt.xlabel("Number of Medals", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.xticks(fontsize=10)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
medal_by_sport = merged_data[merged_data['Medal'].notna()]
medal_by_sport = medal_by_sport.groupby('Sport').size()
Top_10_sport = medal_by_sport.sort_values(ascending=False).head(10)
print(Top_10_sport)

Top_10_sport.plot(
    kind="barh",
    figsize=(12, 8),
    title="Top 10 sport by Medals Records", fontsize=20
)
axsp = plt.gca()
axsp.bar_label(axsp.containers[0])
plt.xlabel("Number of Medals", fontsize=12)
plt.ylabel("Sport", fontsize=12)
plt.xticks(fontsize=10)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
medal_by_year = merged_data[merged_data['Medal'].notna()]
medal_by_year = medal_by_year.groupby('Year').size()
medal_by_year = medal_by_year.sort_index(ascending=False)
print(medal_by_year)

medal_by_year.plot(
    kind="line",
    marker='o',
    figsize=(12, 6),
    title="Medal records by year", fontsize=20
)

plt.xlabel("Year", fontsize=15)
plt.ylabel("Number of Medals", fontsize=15)
plt.xticks(
    ticks=medal_by_year.index[::4],
    rotation=45, fontsize=10)
plt.tight_layout()
plt.show()
```

```python
gender_analysis = merged_data[merged_data['Medal'].notna()]
gender_analysis = gender_analysis.groupby('Sex').size()
gender_analysis = gender_analysis.sort_values(ascending=False)
print(gender_analysis)

gender_analysis.plot(
    kind="bar",
    figsize=(5, 5),
    title="Gender-wise Olympic Medals Records", fontsize=20
)

axg = plt.gca()
axg.bar_label(axg.containers[0])
plt.xlabel("Gender")
plt.ylabel("Number of Medals")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
summer_data = merged_data[
    (merged_data['Medal'].notna()) &
    (merged_data['Year'] >= 1986) &
    (merged_data['Season'] == 'Summer')
]
summer_data = summer_data.groupby(['Year', 'Sex']).size()
print(summer_data)

summer_data= summer_data.unstack()
summer_data.plot(
    kind="line",
    marker='o',
    figsize=(12, 6),
    title="Gender-wise Olympic Medals Records in Summer "
)
plt.xlabel("Year")
plt.ylabel("Number of Medals Records")
plt.xticks(summer_data.index, rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
winter_data = merged_data[
    (merged_data['Medal'].notna()) &
    (merged_data['Year'] >= 1986) &
    (merged_data['Season'] == 'Winter')
]
winter_data = winter_data.groupby(['Year', 'Sex']).size()
print(winter_data)

winter_data= winter_data.unstack()
winter_data.plot(
    kind="line",
    marker='o',
    figsize=(12, 6),
    title="Gender-wise Olympic Medals Records in Winter"
)
plt.xlabel("Year")
plt.ylabel("Number of Medals Records")
plt.xticks(winter_data.index, rotation=45)
plt.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
age_analysis = merged_data[
    (merged_data['Medal'].notna()) &
    (merged_data['Age'].notna())
].copy()
age_analysis['Age_Group'] = pd.cut(
    age_analysis['Age'],
    bins=[10, 20, 30, 40, 50, 100],
    labels=['10-19', '20-29', '30-39', '40-49', '50+'],
    right=False
)
age_group_count = age_analysis.groupby(
    'Age_Group', observed=False
).size()
print(age_group_count)

age_group_count.plot(
    kind="bar",
    figsize=(12, 6),
    title="Medal Records by Athlete Age Group"
)
axa = plt.gca()
axa.bar_label(axa.containers[0])
plt.xlabel("Age_Group")
plt.ylabel("Number of Medals Records")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
